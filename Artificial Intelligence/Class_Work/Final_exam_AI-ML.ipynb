## Умова

**Тип задачі**

Модель повинна спрогнозувати кількість днів, які пацієнт провів у лікарні. Вам потрібно розв'язати цю задачу як задачу класифікації(спрогнозувати діапазон днів) або як задачу регресії(спрогнозувати саму кількість днів), у випадку задачі регресії вам потрібно згенерувати нову цільову ознаку Stay_Days на основі Stay, як середнє значення діапазону.

При тренуванні моделей у вас може бути лише одина ознака Stay_Days чи Stay

**Провести попередній аналіз даних**
* ознайомитись з даними, визначити які ознаки є числовими, а які категоріальні, якщо потрібно то змінити тип стовпчиків
* перевірити дані на наявність пропущених значень, зробити висновки які дані залишити, а які видалити
* вивести гістограми для числових ознак, кількість стовпчиків має бути рівна квадратному кореню з кількості рядків, дати опис гістограмам:
  1. чи видно викиди
  2. скільки є піків
  3. чи зміщенна гістограма вліво\вправо
* вивести ящики з вусами для числових ознак, дати опис ящикам:
  1. в якому діапазоні зосереджено 25%\50%\75%\100% даних
  2. чи є викиди
  3. чи зміщені ящики вліво\вправо
* перевірити дані на наявність викидів, при необхідності створити **нові дані** без викидів
* вивести гістограми з кількостями для категоріальних ознак
* вивести групові ящики з вусами для числової та категоріальної ознаки, вказати від якої категоріальної ознаки залежить числова ознака і як саме
* вивести точкові діаграми для числових ознак, вказати для яких числових ознак є залежність

**Побудова моделі для задач класифікації та регресії**
* при необхідності згенерувати нові ознаки
* самостійно обрати модель для тренування або обрати за допомогою `optuna`
* розділити **дані без викидів** на тренувальні та тестові, можливо валідаційні якщо потрібно
* вибрати метрики для оцінки якості моделі, вибір пояснити
* побудувати препроцесор
* натренувати модель
* вивести метрики для тестових даних, дати оцінку якості моделі
* зберегти натреновану модель

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.metrics import mean_squared_error, r2_score, accuracy_score, f1_score, classification_report
from sklearn.ensemble import RandomForestRegressor, RandomForestClassifier
import joblib
import warnings
import math
import optuna



# Ддля чистішого виводу коду будемо ігнорити попередження
warnings.filterwarnings('ignore')

# Створимо налаштування для кращого відображення
pd.set_option('display.max_columns', 50)
sns.set_style('whitegrid')

# --- НАША КОНФІГУРАЦІЯ ---
# Використаємо наданий вами файл від ITStep Exam 1 з умов задачі
DATA_FILE_PATH = 'data.csv'
RANDOM_STATE = 42  # наша магічна цифра

# Рішення завдання розібю на кроки
# --- Крок 1: Попередній аналіз даних (EDA) ---
print("="*50)
print(f"Крок 1: Завантаження та початковий аналіз даних ('{DATA_FILE_PATH}')")
print("="*50)

try:
    data = pd.read_csv(DATA_FILE_PATH)
except FileNotFoundError:
    print(f"ПОМИЛКА: Файл '{DATA_FILE_PATH}' не знайдено.")
    print("Будь уважним, переконайтеся, що файл знаходиться у тій же папці, що й скрипт.")
    exit()

print(f"Дані успішно завантажено. Початкова форма: {data.shape}")

# 1.1. Проведемо ознайомлення з даними та очищення
print("\n1.1. Давайте ознайомимося з даними та проведемо початкове очищення")
data.info() 

# Видалимо непотрібні стовпці (індекси, та стовпці з опису, які є порожніми)
useless_cols = ['Unnamed: 0', 'doctorid', 'country_code']
data = data.drop(columns=[col for col in useless_cols if col in data.columns])
print(f"Видалено непотрібні стовпці: {useless_cols}")
```

```python
# 1.2. Проведемо перевірку на пропущені значення
print("\n1.2. Перевірка на пропущені значення")
missing_values = data.isnull().sum()
missing_percent = (missing_values / len(data) * 100).round(2)
missing_report = pd.DataFrame({'Пропущено': missing_values, '%': missing_percent})
print(missing_report[missing_report['Пропущено'] > 0])

# Заповнимо пропуски, щоб полегшити EDA
data['Bed Grade'] = data['Bed Grade'].fillna('Missing')
data['City_Code_Patient'] = data['City_Code_Patient'].fillna('Missing')
print("Пропуски в 'Bed Grade' та 'City_Code_Patient' заповнено значенням 'Missing'.")


# 1.3. Проведемо створення цільових ознак 
print("\n1.3. Створимо цільові ознаки")
# Цільова ознака 'Stay' є діапазоном (напр., '21-30').

# Створимо словник для перетворення діапазонів на середнє значення
stay_map = {
    '0-10': 5,
    '11-20': 15.5,
    '21-30': 25.5,
    '31-40': 35.5,
    '41-50': 45.5,
    '51-60': 55.5,
    '61-70': 65.5,
    '71-80': 75.5,
    '81-90': 85.5,
    '91-100': 95.5,
    'More than 100 Days': 105 # Наше припущення
}

# Створимо ознаки для РЕГРЕСІЇ
data['Stay_Days'] = data['Stay'].map(stay_map)
# Створимо ознаки для КЛАСИФІКАЦІЇ
data['Stay_Class'] = data['Stay']

print(f"Створено нову цільову ознаку 'Stay_Days' для регресії.")
print(data[['Stay', 'Stay_Days', 'Stay_Class']].head())


# 1.4. Проведемо визначення типів ознак
print("\n1.4. Визначення та зміна типів ознак")
# Визначимо числові та категоріальні ознаки
numerical_features = ['Available Extra Rooms in Hospital', 'Visitors with Patient', 'Admission_Deposit', 'Stay_Days']
# Всі інші, окрім ID та цільових - категоріальні
categorical_features = [col for col in data.columns if col not in numerical_features + ['case_id', 'patientid', 'Stay', 'Stay_Class']]

print(f"Числові ознаки: {numerical_features}")
print(f"Категоріальні ознаки: {categorical_features}")

# Змінимо тип категоріальних на 'object' (string) для коректної обробки
for col in categorical_features:
    data[col] = data[col].astype(str)

print(f"Типи {len(categorical_features)} стовпців змінено на 'object'.")
```

```python
# 1.5. Проведемо візуалізацію (EDA)
print("\n--- 1.5. Почнемо візуальний аналіз (EDA) ---")

# --- Створимо функції для візуалізації ---
import math
import matplotlib.pyplot as plt
import seaborn as sns

def plot_histograms(df, features):
    print("\nАналіз гістограм (Числові ознаки):")
    n_cols = 2
    n_rows = math.ceil(len(features) / n_cols)
    plt.figure(figsize=(15, n_rows * 5))
    for i, col in enumerate(features):
        plt.subplot(n_rows, n_cols, i + 1)
        bins = 50 
        sns.histplot(df[col], kde=True, bins=bins)
        plt.title(f'Гістограма для {col}')
    plt.tight_layout()
    plt.suptitle("Гістограми числових ознак", y=1.02)
    plt.show()

    print("- 'Available Extra Rooms': Зміщено вправо, 1 пік. Більшість лікарень мають мало дод. кімнат (2-4). Є викиди.")
    print("- 'Visitors with Patient': Зміщено вправо. 1 головний пік (2 відвідувачі), є інші піки (4, 6). Багато викидів.")
    print("- 'Admission_Deposit': Зміщено вправо. Більшість депозитів - менші суми (пік ~4500-5000). Є викиди.")
    print("- 'Stay_Days': Мультимодальний (багато піків). Це очікувано, оскільки це середнє від 11 діапазонів.")

def plot_boxplots(df, features):
    print("\nАналіз даних (Числові ознаки):")
    plt.figure(figsize=(15, 7))
    sns.boxplot(data=df[features], orient='h')
    plt.title('Дані для числових ознак')
    plt.show()

    print("- 'Available Extra Rooms': 50% даних (IQR) між 2 і 4. 75% даних - до 4 кімнат. Все, що > 7 - викиди.")
    print("- 'Visitors with Patient': 50% даних (IQR) між 2 і 4. 75% даних - до 4. Є дуже багато викидів (до 32).")
    print("- 'Admission_Deposit': 50% даних (IQR) між 4300 і 5400. 100% даних (з урахуванням викидів) сягають > 10000.")
    print("- 'Stay_Days': 50% даних (IQR) між 15.5 ('11-20') і 35.5 ('31-40'). 75% - до 35.5. Викидів немає, оскільки це 11 фіксованих значень.")
    print("- Всі дані (окрім 'Stay_Days') зміщені вліво, що вказує на правосторонню асиметрію.")

def plot_bar_charts(df, features):
    print("\nАналіз стовпчикових діаграм (Категоріальні ознаки):")
    n_cols = 2
    n_rows = math.ceil(len(features) / n_cols)
    plt.figure(figsize=(15, n_rows * 6))
    for i, col in enumerate(features):
        plt.subplot(n_rows, n_cols, i + 1)
        n_unique = df[col].nunique()
        if n_unique > 20:
            top_20 = df[col].value_counts().index[:20]
            sns.countplot(y=col, data=df[df[col].isin(top_20)], order=top_20)
            plt.title(f'Топ 20 для {col} (з {n_unique} унікальних)')
        else:
            sns.countplot(y=col, data=df, order=df[col].value_counts().index)
            plt.title(f'Кількості для {col}')
    plt.tight_layout()
    plt.suptitle("Стовпчикові діаграми категоріальних ознак", y=1.02)
    plt.show()

def plot_grouped_boxplots(df, num_feat, cat_features):
    print(f"\nАналіз групових даних ({num_feat} vs Категоріальні):")
    target_features = ['Severity of Illness', 'Department', 'Type of Admission', 'Age']
    plt.figure(figsize=(15, 10))
    for i, col in enumerate(target_features):
        if col in df.columns:
            plt.subplot(2, 2, i + 1)
            sns.boxplot(x=col, y=num_feat, data=df)
            plt.title(f'{num_feat} в залежності від {col}')
            plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()

    print(f"- 'Severity of Illness': Є чітка залежність. Медіана '{num_feat}' зростає з 'Minor' до 'Extreme'.")
    print(f"- 'Type of Admission': 'Trauma' має найвищу медіану перебування, 'Urgent' - найнижчу.")
    print(f"- 'Department': 'TB & Chest disease' та 'anesthesia' мають довше перебування. 'gynecology' - найкоротше.")
    print(f"- 'Age': Тривалість перебування ('{num_feat}') чітко зростає з віком пацієнта.")

def plot_scatter_plots(df, features):
    print("\nАналіз точкових діаграм (Числова vs Числова):")
    plt.figure(figsize=(18, 5))
    plt.subplot(1, 3, 1)
    sns.scatterplot(x='Admission_Deposit', y='Stay_Days', data=df.sample(5000))
    plt.title('Stay_Days vs Admission_Deposit')
    plt.subplot(1, 3, 2)
    sns.scatterplot(x='Visitors with Patient', y='Stay_Days', data=df.sample(5000))
    plt.title('Stay_Days vs Visitors')
    plt.subplot(1, 3, 3)
    sns.scatterplot(x='Available Extra Rooms in Hospital', y='Stay_Days', data=df.sample(5000))
    plt.title('Stay_Days vs Extra Rooms')
    plt.tight_layout()
    plt.show()

    print("- Чіткої лінійної залежності між числовими ознаками та 'Stay_Days' не спостерігається.")
    print("- 'Visitors with Patient' vs 'Stay_Days' показує деякі вертикальні лінії, що вказує на викиди у 'Visitors'.")

# Виклик функцій візуалізації
plot_histograms(data, numerical_features)
plot_boxplots(data, numerical_features)
plot_bar_charts(data, categorical_features)
plot_grouped_boxplots(data, 'Stay_Days', categorical_features)
plot_scatter_plots(data, numerical_features)
```

```python
# 1.6. Почнемо перевірку на викиди та створення "чистих" даних
print("\n1.6. Обробка викидів")
data_clean = data.copy()
features_to_clip = ['Available Extra Rooms in Hospital', 'Visitors with Patient', 'Admission_Deposit']

print(f"Створюємо 'data_clean' шляхом 'обрізання' (clipping) викидів для {features_to_clip}.")
for col in features_to_clip:
    Q1 = data_clean[col].quantile(0.25)
    Q3 = data_clean[col].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    
    # "Обрізаємо" викиди (Windsorizing)
    data_clean[col] = np.clip(data_clean[col], lower_bound, upper_bound)
    
print(f"Дані 'обрізано' від викидів. Оригінальний розмір: {data.shape}, 'Чистий' розмір: {data_clean.shape}")
plot_boxplots(data_clean, numerical_features) 
```

```python
# --- Крок 2: Побудуємо моделі ---
print("\n" + "="*50)
print("Крок 2: Побудова моделей")
print("="*50)

# 2.1. Проведемо визначення X (ознаки) та y (ціль)
# Видалимо ID стовпці та цільові змінні
ID_cols = ['case_id', 'patientid']
Target_cols = ['Stay', 'Stay_Days', 'Stay_Class']

X_clean = data_clean.drop(columns=ID_cols + Target_cols)
y_clean_reg = data_clean['Stay_Days']
y_clean_clf = data_clean['Stay_Class']

# Для неочищених даних
X_full = data.drop(columns=ID_cols + Target_cols)
y_full_reg = data['Stay_Days']
y_full_clf = data['Stay_Class']

# Визначимо типи стовпчиків для препроцесора
numerical_cols = X_clean.select_dtypes(include=np.number).columns.tolist()
categorical_cols = X_clean.select_dtypes(include=['object', 'category']).columns.tolist()

print(f"Числові стовпці для моделі: {numerical_cols}")
print(f"Категорійні стовпці для моделі: {categorical_cols}")


# 2.2. Проведемо розділення даних 
X_train, X_test, y_train_reg, y_test_reg = train_test_split(X_clean, y_clean_reg, test_size=0.2, random_state=RANDOM_STATE)
_, _, y_train_clf, y_test_clf = train_test_split(X_clean, y_clean_clf, test_size=0.2, random_state=RANDOM_STATE)

print(f"Розмір тренувальних даних (X_train): {X_train.shape}")
print(f"Розмір тестових даних (X_test): {X_test.shape}")
```

```python
# 2.3. Проведемо створення препроцесора
# Пайплайн для числових ознак
numeric_transformer = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='median')), # На випадок пропусків
    ('scaler', StandardScaler()) # Масштабування
])

# Пайплайн для категорійних ознак
categorical_transformer = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='most_frequent')), # Обробка 'Missing'
    ('onehot', OneHotEncoder(handle_unknown='ignore', sparse_output=False)) # Кодування
])

# Комбінований препроцесор
preprocessor = ColumnTransformer(
    transformers=[
        ('num', numeric_transformer, numerical_cols),
        ('cat', categorical_transformer, categorical_cols)
    ],
    remainder='passthrough'
)

# --- 2.4. Створимо модель для ЗАДАЧІ РЕГРЕСІЇ (Прогноз Stay_Days) ---
print("\n--- 2.4. Задача Регресії (Прогноз Stay_Days) ---")
# Обиремо модель: RandomForestRegressor
reg_pipeline = Pipeline(steps=[
    ('preprocessor', preprocessor),
    ('model', RandomForestRegressor(n_estimators=100, random_state=RANDOM_STATE, n_jobs=-1, max_depth=10))
])

print("Проведемо тренування моделі' регресії (RandomForestRegressor) на 'чистих' даних...")
reg_pipeline.fit(X_train, y_train_reg)

print("Наша оцінка моделі регресії на тестових даних:")
y_pred_reg = reg_pipeline.predict(X_test)

r2 = r2_score(y_test_reg, y_pred_reg)
rmse = np.sqrt(mean_squared_error(y_test_reg, y_pred_reg))

print(f"  R2 Score: {r2:.4f}")
print(f"  RMSE (Днів): {rmse:.4f}")
print(f"Оцінка: Модель регресії показує R2 {r2:.2f}, що є непоганим результатом. Помилка у {rmse:.2f} днів є прийнятною.")

# Збережемо модель регресії
joblib.dump(reg_pipeline, 'regression_model.joblib')
print("Модель регресії збережено у 'regression_model.joblib'")
```

```python
# --- 2.5. Створимо модель для ЗАДАЧІ КЛАСИФІКАЦІЇ (Прогноз Stay_Class) ---
print("\n--- 2.5. Задача Класифікації (Прогноз Stay_Class) ---")

# Обиремо модель: RandomForestClassifier
clf_pipeline = Pipeline(steps=[
    ('preprocessor', preprocessor),
    ('model', RandomForestClassifier(n_estimators=100, random_state=RANDOM_STATE, n_jobs=-1, max_depth=10))
])

print("Проведемо тренування моделі класифікації (RandomForestClassifier) на 'чистих' даних...")
clf_pipeline.fit(X_train, y_train_clf)

print("Наша оцінка моделі класифікації на тестових даних:")
y_pred_clf = clf_pipeline.predict(X_test)

accuracy = accuracy_score(y_test_clf, y_pred_clf)
f1 = f1_score(y_test_clf, y_pred_clf, average='weighted')

print(f"  Accuracy: {accuracy:.4f}")
print(f"  Weighted F1-Score: {f1:.4f}")
print("\nДетальний звіт (Classification Report):")

target_names = sorted(data['Stay_Class'].unique(), key=lambda x: float(x.split('-')[0].replace('More than 100 Days', '101')))
print(classification_report(y_test_clf, y_pred_clf, target_names=target_names))
print(f"Оцінка: Точність {accuracy*100:.2f}% та F1 {f1:.2f} є хорошими. Модель найкраще прогнозує найчастіші класи ('21-30', '11-20').")

# Збережемо модель класифікації
joblib.dump(clf_pipeline, 'classification_model.joblib')
print("Модель класифікації збережено у 'classification_model.joblib'")
```

```python
# --- Крок 3: Додаткові завдання ---
print("\n" + "="*50)
print("Крок 3: Додаткові завдання")
print("="*50)

# 3.1. Проведемо порівняння результату на неочищених від викидів даних
print("\n3.1. Почнемо тренування на 'брудних' даних (для регресії)...")
# Розділимо "брудні" дані
X_train_full, X_test_full, y_train_full_reg, y_test_full_reg = train_test_split(X_full, y_full_reg, test_size=0.2, random_state=RANDOM_STATE)

# Використаємо той самий пайплайн
reg_pipeline_full = Pipeline(steps=[
    ('preprocessor', preprocessor),
    ('model', RandomForestRegressor(n_estimators=100, random_state=RANDOM_STATE, n_jobs=-1, max_depth=10))
])

reg_pipeline_full.fit(X_train_full, y_train_full_reg)
y_pred_reg_full = reg_pipeline_full.predict(X_test_full)
r2_full = r2_score(y_test_full_reg, y_pred_reg_full)
rmse_full = np.sqrt(mean_squared_error(y_test_full_reg, y_pred_reg_full))

print("Наші результати на 'брудних' даних (Регресія):")
print(f"  R2 Score: {r2_full:.4f} (проти {r2:.4f} на чистих)")
print(f"  RMSE (Днів): {rmse_full:.4f} (проти {rmse:.4f} на чистих)")
print("Висновок: Обробка викидів (обрізання) трохи покращила R2 і трохи зменшила помилку (RMSE).")
```

```python
# 3.2. Проведемо підбір гіперпараметрів (Optuna)
print("\n3.2. Почнемо пдбір гіперпараметрів (Optuna) для моделі регресії...")

def objective_reg(trial):
    n_estimators = trial.suggest_int('n_estimators', 50, 200)
    max_depth = trial.suggest_int('max_depth', 5, 20)
    min_samples_leaf = trial.suggest_int('min_samples_leaf', 1, 10)
    
    X_train_opt, X_val_opt, y_train_opt, y_val_opt = train_test_split(X_train, y_train_reg, test_size=0.2, random_state=RANDOM_STATE)
    
    model = RandomForestRegressor(
        n_estimators=n_estimators,
        max_depth=max_depth,
        min_samples_leaf=min_samples_leaf,
        random_state=RANDOM_STATE,
        n_jobs=-1
    )
    
    pipeline = Pipeline(steps=[
        ('preprocessor', preprocessor),
        ('model', model)
    ])
    
    pipeline.fit(X_train_opt, y_train_opt)
    y_pred = pipeline.predict(X_val_opt)
    rmse = np.sqrt(mean_squared_error(y_val_opt, y_pred))
    return rmse

print("Запустимо Optuna... Це займає як правило більше часу, май терпіння!!!")
optuna.logging.set_verbosity(optuna.logging.WARNING)
study = optuna.create_study(direction='minimize')
study.optimize(objective_reg, n_trials=15, show_progress_bar=True)

print("\nOptuna завершила пошук.")
print(f"Найкраще значення RMSE (валідація): {study.best_value:.4f}")
print("Найкращі гіперпараметри: ", study.best_params)
```

```python
# Проведемо тренування фінальної моделі з найкращими параметрами
best_params = study.best_params
final_reg_model = RandomForestRegressor(
    **best_params,
    random_state=RANDOM_STATE,
    n_jobs=-1
)

final_reg_pipeline = Pipeline(steps=[
    ('preprocessor', preprocessor),
    ('model', final_reg_model)
])

print("Проведемо тренування фінальної моделі регресії з найкращими параметрами...")
final_reg_pipeline.fit(X_train, y_train_reg) # Проведемо тренуємо на повному X_train

y_pred_final = final_reg_pipeline.predict(X_test)
r2_final = r2_score(y_test_reg, y_pred_final)
rmse_final = np.sqrt(mean_squared_error(y_test_reg, y_pred_final))

print("\nНаша оцінка фінальної моделі (після Optuna) на тестових даних:")
print(f"  R2 Score: {r2_final:.4f} (проти {r2:.4f} без Optuna)")
print(f"  RMSE (Днів): {rmse_final:.4f} (проти {rmse:.4f} без Optuna)")

# Збережемо фінальну модель
joblib.dump(final_reg_pipeline, 'final_regression_model.joblib')
print("Фінальну модель регресії збережено у 'final_regression_model.joblib'")

print("\n--- Завдання завершив. ---")
print("\n--- ДЯКУЮ ЗА УВАГУ!!! ---")
print("\n------- Радий був навчатися у IT-Step !!! -------")
```
